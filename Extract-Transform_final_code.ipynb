```python
from pathlib import Path
import json
import pandas as pd

data_dir = Path.cwd()
df = pd.read_csv(data_dir / 'backer_info.csv', encoding='utf-8-sig')
records = df['backer_info'].map(json.loads)
backers_df = pd.DataFrame(records.tolist())
required_columns = ['backer_id', 'cf_id', 'name', 'email']
missing_columns = set(required_columns) - set(backers_df.columns)
if missing_columns:
    raise ValueError(f'Missing fields: {sorted(missing_columns)}')
backers_df = backers_df.loc[:, required_columns].copy()
backers_df['cf_id'] = pd.to_numeric(backers_df['cf_id'], errors='raise').astype('int64')
if backers_df[required_columns].isna().any().any():
    raise ValueError('Backer records must contain all required fields.')
backers_df.head()
```

```python
backers_df[['first_name', 'last_name']] = backers_df['name'].str.strip().str.split(n=1, expand=True).reindex(columns=[0, 1])
backers_df.head()
```

```python
output_path = data_dir / 'backers_transformed.csv'
backers_df.to_csv(output_path, index=False, encoding='utf-8')
print(f'Exported {len(backers_df):,} records to {output_path}')
```
